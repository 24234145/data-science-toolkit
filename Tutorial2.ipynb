```python
#3. Create a numpy array of ones and a numpy array of zeroes

ones_array=np.ones((3,3))
zeros_array=np.zeros((3,3))

print("Array of ones:")
print(ones_array)

print("\nArray of zeros:")
print(zeros_array)
```

```python
#4. Create a random numpy array of size 6x3 and transpose it

random_6x3=np.random.rand(6,3)
transpose_array=random_6x3.T

print("Origina Array (6x3):")
print(random_6x3)

print("\nTransposed Array (3x6):")
print(transpose_array)
```

```python
#5. Create two random numpy array of size n x p and p x m and then compute the dot productimport numpy as np

n=3
p=4
m=2

array_A=np.random.rand(n,p)
array_B=np.random.rand(p,n)
dot_product=np.dot(array_A,array_B)

print("Array A:")
print(array_A)

print("\nArray B:")
print(array_B)

print("\nThis is the dot product:")
print(dot_product)
```
